```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')

reviews = [
    "The product quality is excellent! I am very satisfied with the purchase. Will definitely recommend it.",
    "Terrible customer service. I had to wait two weeks for a response. Not buying again!",
    "Average product. Delivery was on time, but the packaging was damaged."
]
positive_words = ['excellent', 'satisfied', 'recommend', 'good', 'great']
negative_words = ['terrible', 'bad', 'damaged', 'poor', 'worst']
for i,review in enumerate(reviews, 1):
    sentences = sent_tokenize(review)
    print(f"Review: {i}")
    print("Number of sentences:", len(sentences))

    positive_found = []
    negative_found = []

    for sentence in sentences:

        words = word_tokenize(sentence)

        print("Sentence:", sentence)
        print("Number of words:", len(words))

        # Check for positive and negative words
        for w in words:
            word_lower = w.lower()

            if word_lower in positive_words:
                positive_found.append(word_lower)

            if word_lower in negative_words:
                negative_found.append(word_lower)

    print("Positive words:", positive_found)
    print("Negative words:", negative_found)
```
