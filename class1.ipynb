```python
import nltk
from nltk.tokenize import sent_tokenize , word_tokenize
nltk.download('punkt')
nltk.download('punkt_tab')
text="""Bonjour, je m'appelle Marie et j'habite à Paris.
J'aime beaucoup lire des livres et écouter de la musique.
Chaque matin, je prends un café avant d'aller au travail.
Le week-end, je retrouve mes amis dans un petit café.
Nous parlons de nos projets et nous rions beaucoup.
La France est un pays magnifique avec une culture riche."""
sentences=sent_tokenize(text)
for i,j in enumerate(sentences,1):
    print(f"sentence {i}:{j}")
```

```python

for i, j in enumerate(sentences, 1):
    words = word_tokenize(j)
    print(f"Words in sent {i}: {words}")
```

```python
text="""معالجة اللغة الطبيعية (NLP) هي فرع من فروع الذكاء الاصطناعي يساعد الحواسيب على فهم اللغة البشرية ومعالجتها.\
 تُستخدم في مهام مثل تحليل المشاعر، وتصنيف النصوص، والترجمة، وإنشاء روبوتات المحادثة.
 تعتمد معالجة اللغة الطبيعية على تقنيات مثل تقسيم النص إلى كلمات وجمل، واستخراج الكلمات المهمة، وتحليل بنية النص.\
 وتساعد هذه التقنيات الحواسيب على فهم النصوص والتواصل مع البشر بطريقة أكثر طبيعية."""
sentences=sent_tokenize(text)
for i,j in enumerate(sentences,1):
    print(f"sentence {i}:{j}")
for i,j in enumerate(sentences,1):
    words=word_tokenize(j)
    print(f"words in sent. {i}:{words}")
```

```python
import nltk
from nltk.tokenize import sent_tokenize , word_tokenize
nltk.download('punkt')
nltk.download('punkt_tab')
reviews = [ "The product quality is excellent! I am very satisfied with the purchase. Will definitely recommend it.",
    "Terrible customer service. I had to wait two weeks for a response. Not buying again!",
    "Average product. Delivery was on time, but the packaging was damaged."]
positive=["good","excellent","great"]
negative=["bad","worst","damaged"]
print("TOTAL REVIEWS:", len(reviews))
for i,review in enumerate(reviews,1):
    sentences=sent_tokenize(review)
    print(f"REVIEW: {i}")
    print(f"Number of sentences:{len(sentences)}")
    if len(sentences) > 3:
        print(f"flaggeg for review")
    else:
        print(f"normal")
    for j,sentence in enumerate(sentences,1):
        print(f"sentence {j}: {sentence}")
        words=word_tokenize(sentence)
        print(f"words: {words}")
        print(f"no. of words:{len(words)}")
        for word in words:
            if word in positive:
                print("positive:",word)
            if word in negative:
                print("negative:",word)



    


```
