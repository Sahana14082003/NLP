```python

import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Step 1: Download required NLTK data
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

# Step 2: Create the restaurant review dataset
reviews = [
    "The food was absolutely fantastic and the service was great!",
    "The ambiance was okay, but the food was cold and not tasty.",
    "I loved the experience. The staff was friendly and polite."
]

# Step 3: Load English stopwords
stop_words = set(stopwords.words('english'))

# Step 4: Display reviews without removing stopwords
print("STEP 1: REVIEWS WITHOUT STOPWORD REMOVAL\n")

for i, review in enumerate(reviews, start=1):
    words = word_tokenize(review)
    print(f"Review {i}: {words}")

# Step 5: Remove stopwords and punctuation
print("\nSTEP 2: REVIEWS AFTER STOPWORD REMOVAL\n")

cleaned_reviews = []

for i, review in enumerate(reviews, start=1):
    words = word_tokenize(review.lower())

    keywords = [
        word for word in words
        if word not in stop_words and word.isalnum()
    ]

    cleaned_reviews.append(keywords)
    print(f"Review {i}: {keywords}")

# Step 6: Display optimized reviews
print("\nSTEP 3: OPTIMIZED REVIEWS\n")

for i, words in enumerate(cleaned_reviews, start=1):
    print(f"Review {i}: {' '.join(words)}")

# Step 7: Display important words for topic analysis
print("\nSTEP 4: IMPORTANT WORDS FOR TOPIC ANALYSIS\n")

for i, words in enumerate(cleaned_reviews, start=1):
    print(f"Review {i} keywords: {', '.join(words)}")
```
