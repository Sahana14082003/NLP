```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
```

```python
query = "Can you tell me when my package will be delivered to my home?"
token=word_tokenize(query)
print(f"1:{token}")
stop_words=set(stopwords.words('english'))
keywords = []
for word in token:
    if word.lower() not in stop_words:
        keywords.append(word)
print("keywords:", keywords)
```

```python
def optimized_query(query):
  token=word_tokenize(query)
  print(f"1:{token}")
  stop_words=set(stopwords.words('english'))
  stop_words = set(stopwords.words('english'))
  custom_stopwords = {'customer', 'order', 'product'}
  stop_words.update(custom_stopwords)
  keywords = []
  for word in token:
      if word.lower() not in stop_words:
          keywords.append(word)
  return keywords
user_query = input("Enter your query: ")
oq = optimized_query(user_query)
print("Optimized Query:", oq)
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

review = [
    "The food was absolutely fantastic and the service was great!",
    "The ambiance was okay, but the food was cold and not tasty.",
    "I loved the experience. The staff was friendly and polite."
]
for r in review:
  token=word_tokenize(r)
  stop_words=set(stopwords.words('english'))
  keywords = []
  for word in token:
      if word.lower() not in stop_words:
          keywords.append(word)
  print("Original Review:", r)
  print("Keywords:", keywords)
  print()
```

```python
import re
import nltk

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

nltk.download('punkt_tab')
nltk.download('stopwords')

posts = [
    "OMG! I love this product 😍!!! Check it out at https://example.com @company #AmazingProduct",
    "I am disappointed with the service 😡. Visit www.example.com #BadService @brand",
    "Great experience! 😊 The product is amazing! #HappyCustomer"
]

stop_words = set(stopwords.words('english'))

for post in posts:

    # Step 1: Remove URLs
    post = re.sub(r'https?://\S+|www\.\S+', '', post)

    # Step 2: Remove hashtags and mentions
    post = re.sub(r'#[\w]+|@[\w]+', '', post)

    # Step 3: Convert to lowercase
    post = post.lower()

    # Step 4: Remove special characters and punctuation
    post = re.sub(r'[^a-zA-Z\s]', '', post)

    # Step 5: Tokenization
    token = word_tokenize(post)

    # Step 6: Remove stopwords
    keywords = []

    for word in token:
        if word not in stop_words:
            keywords.append(word)

    print("Original Post:", post)
    print("Clean Text:", keywords)
    print()
```
